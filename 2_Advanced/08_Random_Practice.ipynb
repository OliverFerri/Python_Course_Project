```python
# Importing Libraries
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df_filtered_salary = df[pd.notna(df['salary_year_avg'])].copy()
```

```python
def projected_salary(salary):
    return salary * 1.03
```

```python
df_filtered_salary['projected_salary_year'] = df_filtered_salary['salary_year_avg'].apply(projected_salary)
```

```python
df_filtered_salary['salary_inflated'] = df_filtered_salary['salary_year_avg'].apply(lambda salary: salary * 1.03)
```

```python
df_filtered_salary[['salary_year_avg', 'projected_salary_year', 'salary_inflated']]
```

```python
import ast

ast.literal_eval(df['job_skills'][1])
```

```python
def clean_list(skill_list):
    if pd.notna(skill_list):
        return ast.literal_eval(skill_list)

df['job_skills'] =  df['job_skills'].apply(clean_list)
```

```python
df['job_skills'] =  df['job_skills'].apply(lambda skill_list: ast
                                           .literal_eval(skill_list) 
                                           if pd.notna(skill_list) else skill_list)
```

```python
def projected_inflation(row):
    if "Senior" in row['job_title_short']:
        return 1.05 * row['salary_year_avg']
    else:
        return 1.03 * row['salary_year_avg']
```

```python
df_filtered_salary['salary_inflated'] = df_filtered_salary.apply(projected_inflation, axis=1)
```

```python
df_filtered_salary[['job_title_short', 'salary_year_avg', 'salary_inflated']].sample(20)
```

```python
# Lambda Function Version
df_filtered_salary['salary_year_inflated'] = df_filtered_salary.apply(
    lambda row: 1.05 * row['salary_year_avg'] if 'Senior' in row['job_title_short']
    else 1.03 * row['salary_year_avg'],
    axis=1
)
```

```python
df_filtered_salary[['job_title_short', 'salary_year_avg', 'salary_year_inflated']].sample(20)
```
