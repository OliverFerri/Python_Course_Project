```python
# Importing Libraries
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df_us_jobs = df[df['job_country'] == 'United States'].copy()
```

```python
df_us_jobs
```

```python
df_us_jobs['job_posted_month'] = df_us_jobs['job_posted_date'].dt.strftime('%B')
```

```python
df_us_jobs
```

```python
df_us_pivot = df_us_jobs.pivot_table(
    index='job_posted_month', 
    columns='job_title_short',
    aggfunc='size'
)
```

```python
df_us_pivot
```

```python
df_us_pivot.reset_index(inplace=True)
```

```python
df_us_pivot['month_no'] = pd.to_datetime(df_us_pivot['job_posted_month'], format='%B').dt.month
df_us_pivot['month_no']
```

```python
df_us_pivot.sort_values('month_no', inplace=True)
df_us_pivot
```

```python
df_us_pivot.set_index('job_posted_month', inplace=True)
```

```python
df_us_pivot.drop(columns=['month_no'], inplace=True)
```

```python
top_3 = df_us_jobs['job_title_short'].value_counts(ascending=False).head(3)
```

```python
top_3 = top_3.index.tolist()
```

```python
df_us_pivot[top_3].plot(kind='line')

plt.ylabel('Job Count')
plt.xlabel('2023')
plt.title('Job Count per Month')
plt.xticks(rotation=-45, ha='left')
plt.show()
```
