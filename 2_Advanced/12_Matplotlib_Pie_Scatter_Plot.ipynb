```python
# Importing Libraries
import ast
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['job_skills'] = df['job_skills'].apply(
    lambda x: ast.literal_eval(x) if pd.notna(x) else x
)
```

```python
df['job_work_from_home'].value_counts().plot(kind='pie', startangle=90, autopct='%1.1f%%', fontsize=13)

plt.title('Work From Home Satus')
plt.ylabel('')
plt.show()
```

```python
dict_column = {
    'job_work_from_home': 'Work from Home',
    'job_no_degree_mention': 'Job Degree Req.', 
    'job_health_insurance': 'Health Insurance Offered'
}

for key, value in dict_column.items():
    fig, ax = plt.subplots()
    data = df[key].value_counts()
    ax.pie(data, labels=data.index, autopct='%1.1f%%', startangle=90)
    ax.set_title(value)

plt.ylabel('')
plt.show()
```

```python
fig, ax = plt.subplots(1, 3)

dict_column = {
    'job_work_from_home': 'Work from Home',
    'job_no_degree_mention': 'Job Degree Req.', 
    'job_health_insurance': 'Health Insurance Offered'
}

for i, (col, title) in enumerate(dict_column.items()):
    data = df[col].value_counts()
    ax[i].pie(data, labels=data.index, autopct='%1.1f%%', startangle=90)
    ax[i].set_title(title)

fig.suptitle('Pie Charts', fontsize=15)
fig.tight_layout()
```
